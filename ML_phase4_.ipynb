```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
import seaborn as sns
```

```python
sns.set_theme(style="whitegrid")
```

```python
df = pd.read_csv("/content/loan_data_new.csv")
```

```python
plt.figure(figsize=(7, 4))
sns.countplot(x="Credit History", hue="Loan Status", data=df, palette="Set1")
plt.title("Loan Approval Status by Credit History")
plt.xlabel("Credit History (1.0 = Good, 0.0 = Poor)")
plt.ylabel("Applicant Count")
plt.legend(title="Loan Status", labels=["Rejected", "Approved"])
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(7, 4))
sns.boxplot(x="Loan Status", y="Person Income", data=df, palette="Set2")
plt.title("Applicant Income Distribution by Loan Status")
plt.xlabel("Loan Status")
plt.ylabel("Applicant Income")
plt.ylim(0, 25000)  # Filter extreme outliers for clear view
plt.tight_layout()
plt.show()
```

```python
df_encoded = df.copy()
for col in df_encoded.select_dtypes(include=["object"]).columns:
    df_encoded[col] = df_encoded[col].astype("category").cat.codes

plt.figure(figsize=(10, 6))
sns.heatmap(
    df_encoded.corr(),
    annot=True,
    fmt=".2f",
    cmap="coolwarm",
    linewidths=0.5,
)
plt.title("Correlation Matrix Heatmap")
plt.tight_layout()
plt.show()
```
