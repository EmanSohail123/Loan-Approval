```python
import numpy as np
import pandas as pd
```

```python
df=pd.read_csv("/loan_data_new.csv")
```

```python
print(df.isnull().sum())
```

```python
duplicate_count=df.duplicated().sum()
print(duplicate_count)
if duplicate_count>0:
  df.drop_duplicates(inplace=True)
if "Loan Status" in df.columns:
    df["Loan Status"] = df["Loan Status"].map({"Y": 1, "N": 0})
```

```python
if "Dependents" in df.columns and df["Dependents"].dtype == "object":
    df["Dependents"] = df["Dependents"].replace("3+", 3).astype(int)
```

```python
binary_mappings = {
    "Gender": {"Male": 1, "Female": 0},
    "Married": {"Yes": 1, "No": 0},
    "Education": {"Graduate": 1, "Not Graduate": 0},
    "Self_Employed": {"Yes": 1, "No": 0},
}
```

```python
for col, mapping in binary_mappings.items():
    if col in df.columns and df[col].dtype == "object":
        df[col] = df[col].map(mapping)

# One-Hot Encoding for multi-category nominal features (e.g., Property_Area)
if "Property_Area" in df.columns:
    df = pd.get_dummies(df, columns=["Property_Area"], drop_first=True, dtype=int)
```
